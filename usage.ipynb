Import the `cplot` module for plotting together with a sample function $f_0 : \mathbb C \rightarrow \mathbb C$ from the `param` module.

```python
import cplot
from param import f0
```

The sample function $f_0 : \mathbb C \rightarrow \mathbb C$ takes exactly one complex argument.

```python
f0(1 + 1j)
```

Define the region you want to plot and invoke `complex_plot` from the `cplot` module.

```python
# Real numbers range
re_range = (-1e-4, 1e-4)
# Imaginary numbers range
im_range = (-1e-4, 1e-4)
cplot.complex_plot(f0, re_range, im_range)
```
